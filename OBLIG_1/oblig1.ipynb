```python
"""
Graident decent algorithm

"""
def gradient_descent(X_b, y, eta, n_epochs):
    m = len(X_b) # nr of samples
    n = X_b.shape[1]  # nr of parameters (features + bias)

    np.random.seed(42)
    theta = np.random.randn(n, 1)  # randomly initlizes each paramter

    for epoch in range(n_epochs):
        gradients = 2 / m * X_b.T @ (X_b @ theta - y) # compute gradient, see page 142
        theta = theta - eta * gradients # compute theta for next iteration
    return theta
```

```python
from sklearn.preprocessing import StandardScaler

feature = 'gmean_fie'
X_single = X[[feature]] # select the feature as dataframe
y_target = y['critical_temp']

scaler = StandardScaler() # scalar
scaled_X = scaler.fit_transform(X_single) # scale the feature

X_b = np.c_[np.ones((len(scaled_X), 1)), scaled_X] # get 2 x n, from two (nx1) matrice, with 1 column of 1s. This is to handle the constant c_0 term.

y_col = y_target.values.reshape(-1, 1) # create 2d array from 1d y_target. It makes it 1 column and n rows.

# this is needed as matmul need dims of (2xn)(nx1) to get the (2x1) result we want

theta_weak = gradient_descent(X_b, y_col, 0.1, 1000)
print(theta_weak)
# here theta[0] is b, and theta[1] is m for the line mx+b
```

```python
# we can make this a function for further use:
def linear_regression_GD(feature, target, X, y, eta:float, epoch):

  X_single = X[[feature]]
  y_target = y[target]

  scaler = StandardScaler()
  scaled_X = scaler.fit_transform(X_single)

  X_b = np.c_[np.ones((len(scaled_X), 1)), scaled_X]
  y_col = y_target.values.reshape(-1, 1)

  theta = gradient_descent(X_b, y_col, eta, epoch)

  return theta, X_b, scaled_X, X_single
```

Standardizing the features is important to keep the gradient value small, so we dont overshoot.

It keeps our values resonable. It also makes the cost area more "bowl" like which we want for gradient decent.

It doesnt let large scales dominate, as we now are working with standard deviations for each feature.


We get the result that critical_temp = -0.8599 * scaled_gmean_fie + 34.4212, scaled_gmean_fie = ((gmean_fie-mean)/std). This mean change per standard deviaiton of gmean_fie.

```python
from sklearn.model_selection import train_test_split


def linear_regression_GD_tt_split(feature, target, X, y, eta:float, epoch, test, i):

  X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=test, random_state=i) # random_state = i, for reproducable result across calls, ie split for i=1 is always the same. However i=2 gives different split.


  X_single_train = X_train[[feature]]
  X_single_test = X_test[[feature]]


  y_target_train = y_train[target]
  y_target_test = y_test[target]


  scaler = StandardScaler()
  scaled_X_train = scaler.fit_transform(X_single_train) # fit + transform on train
  scaled_X_test = scaler.transform(X_single_test) # transform on test


  X_b_train = np.c_[np.ones((len(scaled_X_train), 1)), scaled_X_train]
  X_b_test = np.c_[np.ones((len(scaled_X_test), 1)), scaled_X_test]


  y_col_train = y_target_train.values.reshape(-1, 1)
  y_col_test = y_target_test.values.reshape(-1, 1)


  theta = gradient_descent(X_b_train, y_col_train, eta, epoch)

  return theta, X_b_test, y_col_test # theta contains the parameters for the regression.
```

```python
# evaluate MSE, RMSE and R^2
# from sklearn.metrics import r2_score, mean_squared_error

def get_metrics(X_b, y, theta):
  y_pred = X_b @ theta # create the regression line

  mse = np.round(np.mean((y - y_pred)**2), 4)
  rmse = np.round(np.sqrt(mse), 4)

  res_ss = np.sum((y - y_pred)**2)
  tot_ss = np.sum((y - np.mean(y))**2)
  r2 = np.round(1 - (res_ss/tot_ss), 4)

  return {'mse': mse, 'rmse': rmse, 'r2': r2}
```

```python
def multiple_linear_regression(target, X, y, eta:float, epoch, test, i):
  X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=test, random_state=i) # random_state = i, for reproducable result across calls, ie split for i=1 is always the same. However i=2 gives different split.



  y_target_train = y_train[target]
  y_target_test = y_test[target]


  scaler = StandardScaler()
  scaled_X_train = scaler.fit_transform(X_train) # fit + transform on train
  scaled_X_test = scaler.transform(X_test) # transform on test


  X_b_train = np.c_[np.ones((len(scaled_X_train), 1)), scaled_X_train]
  X_b_test = np.c_[np.ones((len(scaled_X_test), 1)), scaled_X_test]


  y_col_train = y_target_train.values.reshape(-1, 1)
  y_col_test = y_target_test.values.reshape(-1, 1)


  theta = gradient_descent(X_b_train, y_col_train, eta, epoch)

  return theta, X_b_test, y_col_test # theta contains the parameters for the regression.
```

```python
### NTNU GPT
def get_pair_corr(f1, f2, corr_pairs):
    """Look up correlation between two features from a MultiIndex pairs Series,
    checking both orderings since (f1, f2) and (f2, f1) are the same thing."""
    if (f1, f2) in corr_pairs.index:
        return corr_pairs[(f1, f2)]
    elif (f2, f1) in corr_pairs.index:
        return corr_pairs[(f2, f1)]
    return 0.0  # not found = assume no strong correlation on record

def select_low_multicollinearity_high_target_corr(corr_with_target, corr_pairs,
                                                     n_features=10,
                                                     multicollin_threshold=0.85,
                                                     candidate_pool=30):
    # Stage 1: candidate pool ranked by |correlation| with target
    ranked = corr_with_target.abs().sort_values(ascending=False)
    candidates = ranked.head(candidate_pool).index.tolist()

    selected = []
    log = []

    for feature in candidates:
        if len(selected) == n_features:
            break

        conflict = None
        for chosen in selected:
            pc = abs(get_pair_corr(feature, chosen, corr_pairs))
            if pc > multicollin_threshold:
                conflict = (chosen, pc)
                break

        if conflict is None:
            selected.append(feature)
            log.append((feature, ranked[feature], "SELECTED", None))
        else:
            log.append((feature, ranked[feature], "SKIPPED", conflict))

    log_df = pd.DataFrame(log, columns=["feature", "target_corr", "status", "conflict_with"])
    return selected, log_df

# run it
final_10, selection_log = select_low_multicollinearity_high_target_corr(
    corr_with_target, corr_pairs, n_features=10, multicollin_threshold=0.85, candidate_pool=30
)

print("Final 10 selected features:")
print(final_10)
# these features should have a high correlation with the target, and there should be little multi-collinearity within the features. These will be used for the polynomial regression.
```

```python
from sklearn.linear_model import LinearRegression, RidgeCV
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import r2_score
def polynomial_regression(features: list, target: str, X, y, rs: int):
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, train_size=0.8, random_state=rs
    )

    X_features_train = X_train[features]
    X_features_test = X_test[features]

    y_train_target = y_train[target]
    y_test_target = y_test[target]

    # scale
    scaler = StandardScaler()
    scaled_X_train = scaler.fit_transform(X_features_train)
    scaled_X_test = scaler.transform(X_features_test)

    # polynomial features — fit once, transform both
    poly = PolynomialFeatures(2, include_bias=False) # get rows of type [1, x, x^2], where x is a feature
    X_poly_train = poly.fit_transform(scaled_X_train)
    X_poly_test = poly.transform(scaled_X_test)

    # fit model
    model = LinearRegression()
    model.fit(X_poly_train, y_train_target)

    # predict + evaluate
    y_pred_train = model.predict(X_poly_train)
    y_pred_test = model.predict(X_poly_test)

    train_r2 = r2_score(y_train_target, y_pred_train)
    test_r2 = r2_score(y_test_target, y_pred_test)


    return {
        "model": model,
        "poly": poly,
        "scaler": scaler,
        "train_r2": train_r2,
        "train_r2": test_r2,
        "y_pred_test": y_pred_test,
    }
```

The features with low relevance will have less impact, as they have smaller coefficients. This generalizes better because the features that are not relevant are not "weighted as much, as the L2 penalty will reduce the overall magnitude of all coefficients.


```python
from sklearn.linear_model import LassoCV

def lasso_reg(X, y, test_size=0.2, random_state=42):
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=test_size, random_state=random_state
    )

    alphas = np.logspace(-3, 3, 20)
    pipe = Pipeline([('scaler', StandardScaler()), ('lasso', LassoCV(alphas=alphas, cv=5, max_iter=10000))])

    pipe.fit(X_train, y_train)
    best_alpha = pipe.named_steps['lasso'].alpha_
    coefs = pipe.named_steps['lasso'].coef_
    intercept = pipe.named_steps['lasso'].intercept_

    names = X.columns if hasattr(X, 'columns') else [f'X{i}' for i in range(X.shape[1])]
    idxs = np.argsort(np.abs(coefs))[::-1]


    print("Best alpha:", best_alpha)
    print("Intercept:", intercept)
    for i in idxs:
      print(f"{names[i]}: {coefs[i]:.4f}")

    # evaluate on held-out test set
    y_pred = pipe.predict(X_test)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, y_pred)

    print(f"\nTest MSE: {mse:.4f}")
    print(f"Test RMSE: {rmse:.4f}")
    print(f"Test R2: {r2:.4f}")

    # return pipe, X_test, y_test, y_pred
```

```python
lasso_reg(X,y_target)
```

Comparing with MLR we get better performance, lower MSE and higher R^2. Here we get more obvious regularization as it drives some of the one of the multicolinear features to 0. This is a type of feature selection, it selects highly corrolated, non-multicolinear features. This effectivly reduces the dimention of our input space.

Comparing with the multicolinear pairs in Q1.2.3. We can see that, all of the pairs with high multicolinearity have atleast one feature with low or near zero coeffeient.
